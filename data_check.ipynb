```python
import os
import pandas as pd

li = os.listdir('./data/')
print(len(li))
datasets = pd.DataFrame()
for i in li:
    dataset = pd.read_pickle('./data/'+i)
    datasets = pd.concat([datasets, dataset])

datasets.shape
```

```python
db = datasets['리뷰'].str.rsplit("\n", expand=True)
db.columns = ['0', '별점', '리뷰1', '0', '작성자', '0', '작성일', '0', '0', '추천 반대', '0', '0', '0']
db['리뷰1'] = db['리뷰1'].str.replace('관람객 ','')

db['추천'] = db['추천 반대'].str.rsplit(" ", expand=True)[0]
db['반대'] = db['추천 반대'].str.rsplit(" ", expand=True)[1]

db = db[[ '별점', '리뷰1',  '작성자',  '작성일',  '추천', '반대']]

datasets.reset_index(drop=True, inplace=True)
db.reset_index(drop=True, inplace=True)
df_all = pd.concat([datasets, db], axis=1)

df_all = df_all.dropna(subset=['리뷰1'])
df_all = df_all[-df_all['리뷰1'].str.contains('스포일러')]
df_all = df_all[df_all['리뷰1'] != '관람객']
df_all = df_all.drop_duplicates().reset_index(drop=True)
df_all.shape
```
