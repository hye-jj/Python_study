```python
# 한국어 텍스트 입력
text = "안녕하세요, 저는 인공지능입니다."

# 텍스트를 로마자로 변환
romanized_text = unidecode(text)

# 텍스트를 토큰화
inputs = tokenizer(romanized_text, return_tensors="pt")

# 입력 텐서를 LongTensor로 변환
inputs = {key: value.long() for key, value in inputs.items()}

# 모델을 사용하여 오디오 생성
with torch.no_grad():
    outputs = model(**inputs).waveform

# 오디오 데이터를 numpy 배열로 변환
audio_numpy = outputs.squeeze().cpu().numpy()

# 샘플링 레이트 (모델에 따라 다를 수 있음)
sample_rate = model.config.sampling_rate

# 오디오 파일로 저장
output_path = "output_audio.wav"
sf.write(output_path, audio_numpy, sample_rate)

print(f"Audio saved to {output_path}")

# IPython Audio 객체를 사용하여 오디오 재생
audio_obj = Audio(audio_numpy, rate=sample_rate)
audio_obj
```

```python
from melo.api import TTS

# Speed is adjustable
speed = 1.0
device = 'cpu' # or cuda:0

# Set the directory where the model should be stored
model_dir = "./saved_model2"
text = "안녕하세요! 오늘은 날씨가 정말 좋네요."
model = TTS(language='KR', device=device, cache_dir=model_dir)
speaker_ids = model.hps.data.spk2id

output_path = 'kr.wav'
model.tts_to_file(text, speaker_ids['KR'], output_path, speed=speed)
```

```python
from melo.api import TTS

# Assuming you've manually downloaded the model to a specific directory
model_dir = "./saved_model2"
model = TTS(language='KR', device='cpu')
model.load_model_from_directory(model_dir)

text = "안녕하세요! 오늘은 날씨가 정말 좋네요."
speaker_ids = model.hps.data.spk2id

output_path = 'kr.wav'
model.tts_to_file(text, speaker_ids['KR'], output_path, speed=speed)
```
